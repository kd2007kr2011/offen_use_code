```python
import os
path = "C:\\Users\\86139\\Desktop\\要发邮件"
file_list = os.listdir(path)
import re
reg_name = re.compile('[^22信科2班]+')
import pandas as pd
mail_data =pd.read_excel('C:\\Users\\86139\\Desktop\\杂七杂八\\个人邮箱11.xlsx')
mail_dic = {}
for i in range(mail_data.shape[0]):
    mail_dic[mail_data.iloc[i,0]] =mail_data.iloc[i,1]

import smtplib
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.base import MIMEBase
from email import encoders
from email.header import make_header
import time
```

```python
send_list = []
```

```python
for file in file_list:
    a = file.split('-')[0]
    name = reg_name.findall(a)[0]
    file_path = os.path.join(path,file)
    mail = mail_dic[name]
    print("正在给",name,"发送邮件")
        # 设置发件人、收件人和主题
    sender_email = "1007622827@qq.com"
    receiver_email = mail
    subject = "Java第7、8章平时作业"

    # 创建邮件内容
    msg = MIMEMultipart()
    msg['From'] = sender_email
    msg['To'] = receiver_email
    msg['Subject'] = subject

    # 添加正文
    body = "下载后可看到老师批改的痕迹，有什么问题请联系我"
    msg.attach(MIMEText(body, 'plain'))

    # 添加附件
    filename = file  # 修改为您的Word文件名（.doc文件）

    attachment = open(file_path, "rb")  # 修改为您Word文件的路径
    part = MIMEBase('application', 'msword')  # 修改为正确的内容类型，表示Word文件
    part.set_payload(attachment.read())
    encoders.encode_base64(part)
    encoded_filename = make_header([(filename, None)]).encode('utf-8')
    part.add_header('Content-Disposition', f"attachment; filename={encoded_filename}")
    msg.attach(part)

    #连接到SMTP服务器并发送邮件
    if name == '李卓钊':
        server = smtplib.SMTP('smtp.qq.com', 587)
        server.starttls()
        server.login("1007622827@qq.com", "gdretkelkmvtbajh")
        text = msg.as_string()
        server.sendmail(sender_email, receiver_email, text)
        server.quit()
        time.sleep(15)
#     if name not in send_list:
#         server = smtplib.SMTP('smtp.qq.com', 587)
#         server.starttls()
#         server.login("1007622827@qq.com", "oorhqemicrvwbcei")
#         text = msg.as_string()
#         server.sendmail(sender_email, receiver_email, text)
#         server.quit()
#         send_list.append(name)
#         time.sleep(10)
```

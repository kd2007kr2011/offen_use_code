```python
from selenium import webdriver
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.keys import Keys
import requests
import os
#函数定义
dic = {
    '6-2':'6-2非平衡电桥测量热敏电阻的温度系数',
    '6-5':'6-5迈克尔逊干涉仪测量激光波长',
    '6-6':'6-6掠入射法测量三棱镜折射率',
    '7-5':'7-5利用双光栅测量微弱振动'
}
def down_pic(down_list):
    for i in range(len(down_list)):
        response = requests.get(url = down_list[i])
        response.encoding= 'utf-8'
        data = response.content
        name = '图片{}.jpg'.format(i)
        with open(name, 'wb') as fp:
            fp.write(data)
        print(name,'已经下载成功')
```

```python
#图像更改的代码
from PIL import Image
import os
def img_2_pdf(in_path, out_path):
    files = os.listdir(in_path)
    img_files = []
    sources = []
    for file in files:
        if 'png' in file or 'jpg' in file or 'jpeg'in file:
            img_files.append(in_path+'\\'+ file)
    img_files.sort()
    output = Image.open(img_files[0])
    width,height = output.size
    if width>height:
        output = output.rotate(-90,expand=True)
        output = output.crop(output.getbbox())
    for file in img_files:
        img_file = Image.open(file)
        width,height = img_file.size
        if width>height:
            img_file = img_file.rotate(-90,expand=True)
            img_file = img_file.crop(img_file.getbbox())
        if img_file.mode == "RGBA":
            img_file = img_file.convert("RGB")
        sources.append(img_file)
    sources.pop(0)
    output.save(out_path, "pdf", save_all=True, append_images=sources)

#创建名单的字典
import pandas as pd
data = pd.read_excel("C:\\Users\\86139\\Desktop\\杂七杂八\\22信科学号.xlsx")
name_numble = {}
for i in range(data.shape[0]):
    name = data.iloc[i,0]
    numble = data.iloc[i,1]
    name_numble[name] = numble
change_list = ['刘璐璐','罗方静','吴一鸣','吴佳新']
name_list = data.iloc[:,0].values.tolist()
```

```python
path = 'C:\\Users\\86139\\Desktop\\大学物理实验'
pdf_list = os.listdir(path)
name = input("请输入姓名:")
for file_name in pdf_list:
    img_path = os.path.join(path,file_name)
    pdf_name = str(name_numble[name]) + '-'+name + '-'+ file_name + '.pdf'
    out_path = os.path.join(img_path,pdf_name)
    img_2_pdf(img_path,out_path)
```

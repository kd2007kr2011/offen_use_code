```python
import fitz  # PyMuPDF  
from PIL import Image  

def images_to_pdf(images, output_pdf_path):  
    # 创建一个PDF文档  
    doc = fitz.open()  
  
    # 遍历图片列表  
    for img_path in images:  
        # 使用Pillow打开图片  
        img = Image.open(img_path)  
          
        # 确保图片是以RGB模式打开的（有些图片可能是CMYK或灰度模式）  
        if img.mode != 'RGB':  
            img = img.convert('RGB')  
          
        # 将Pillow图片保存为字节流中的临时PDF  
        # 注意：这里使用Pillow将图片转换为PDF可能不是最高效的方式，但它简单易行  
        img_byte_arr = io.BytesIO()  
        img.save(img_byte_arr, "PDF", resolution=100.0)  
        img_byte_arr.seek(0)  
          
        # 从字节流中加载PDF  
        img_pdf = fitz.open(stream=img_byte_arr)  
          
        # 将图片PDF的第一页添加到主PDF文档中  
        page = img_pdf.load_page(0)  
        doc.insert_page(page)  
          
        # 关闭临时PDF  
        img_pdf.close()  
  
    # 保存PDF文件  
    doc.save(output_pdf_path)  
    doc.close()  
  
# 图片文件夹路径和输出PDF路径  
images_folder = "C:\\Users\\86139\\Desktop\\教材"
images = [f'{images_folder}/{img}' for img in os.listdir(images_folder) if img.lower().endswith(('.png', '.jpg', '.jpeg'))]  
output_pdf_path = 'output.pdf'  
  
# 确保导入了os和io模块  
import os  
import io  
  
# 调用函数  
images_to_pdf(images, output_pdf_path)
```
